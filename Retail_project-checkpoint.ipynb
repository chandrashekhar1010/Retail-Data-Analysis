```python
import pandas as pd
df = pd.read_csv('online_retail.csv')
df.info()
df.isnull().sum()
```

```python
df = df.dropna(subset=['order_id', 'customer_id']).copy()
df.loc[:, 'age'] = df['age'].fillna(df['age'].median())
df.isnull().sum()
```

```python
df['order_date'] = pd.to_datetime(
    df['order_date'],
    dayfirst=True,
    errors='coerce'
)
df['age'] = df['age'].astype(int)
print("Unparseable dates:", df['order_date'].isnull().sum())
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_day_name'] = df['order_date'].dt.day_name()
df['age_group'] = pd.cut(df['age'], bins=[0,18,25,35,45,60,100],
                          labels=['<18','18-25','26-35','36-45','46-60','60+'])
```

```python
df['gender'] = df['gender'].str.strip().str.lower()
gender_map = {'male': 'Male', 'm': 'Male', 'female': 'Female', 'f': 'Female'}
df['gender'] = df['gender'].map(gender_map).fillna('Other')
print(df['gender'].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")
os.makedirs('images', exist_ok=True)
```

```python
category_counts = df['product_category'].value_counts()
plt.figure(figsize=(8,5))
sns.barplot(x=category_counts.values, y=category_counts.index, palette='viridis')
plt.title('Orders by Product Category')
plt.xlabel('Number of Orders')
plt.ylabel('Category')
plt.savefig('images/orders_by_category.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
region_counts = df['region'].value_counts()
plt.figure(figsize=(8,5))
sns.barplot(x=region_counts.index, y=region_counts.values, palette='mako')
plt.title('Orders by Region')
plt.xticks(rotation=45)
plt.savefig('images/orders_by_region.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(5,5))
df['gender'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90,
                                       colors=['#4C72B0','#DD8452','#55A868'])
plt.title('Orders by Gender')
plt.ylabel('')
plt.savefig('images/orders_by_gender.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='age_group', order=['<18','18-25','26-35','36-45','46-60','60+'], palette='coolwarm')
plt.title('Orders by Age Group')
plt.savefig('images/orders_by_age_group.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
yearly_orders = df.groupby('order_year').size()
plt.figure(figsize=(8,5))
yearly_orders.plot(kind='line', marker='o')
plt.title('Number of Orders per Year')
plt.xlabel('Year')
plt.ylabel('Orders')
plt.savefig('images/orders_per_year.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# --- Chart 2: Orders by Region ---
region_counts = df['region'].value_counts()
plt.figure(figsize=(8,5))
sns.barplot(x=region_counts.index, y=region_counts.values, palette='mako')
plt.title('Orders by Region')
plt.xticks(rotation=45)
plt.savefig('images/orders_by_region.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
top_combo = df.groupby(['region','product_category']).size().reset_index(name='count')
pivot = top_combo.pivot(index='region', columns='product_category', values='count').fillna(0)
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='YlGnBu')
plt.title('Product Category Popularity by Region')
plt.show()
```
